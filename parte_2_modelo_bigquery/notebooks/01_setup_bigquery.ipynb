```python
from google.cloud import bigquery
from google.oauth2 import service_account
import os
from dotenv import load_dotenv

load_dotenv()

credentials = service_account.Credentials.from_service_account_file(
    os.getenv("GOOGLE_APPLICATION_CREDENTIALS")
)
client = bigquery.Client(
    project=os.getenv("GCP_PROJECT_ID"),
    credentials=credentials
)

print("Conectado correctamente al proyecto:", client.project)
```

```python
dataset_id = f"{client.project}.{os.getenv('BQ_DATASET_ID')}"
dataset = bigquery.Dataset(dataset_id)
dataset.location = "EU"

dataset = client.create_dataset(dataset, exists_ok=True)
print(f"Dataset {dataset_id} listo")
```

```python
schema_customers = [
    bigquery.SchemaField("customer_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("first_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("last_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("email", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("country", "STRING", mode="NULLABLE"),
    bigquery.SchemaField("city", "STRING", mode="NULLABLE"),
    bigquery.SchemaField("acquisition_channel", "STRING", mode="NULLABLE"),
    bigquery.SchemaField("registration_date", "DATE", mode="REQUIRED"),
]
table = bigquery.Table(f"{dataset_id}.customers", schema=schema_customers)
client.create_table(table, exists_ok=True)
print("Tabla customers creada")
```

```python
schema_categories = [
    bigquery.SchemaField("category_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("description", "STRING", mode="NULLABLE"),
]
table = bigquery.Table(f"{dataset_id}.categories", schema=schema_categories)
client.create_table(table, exists_ok=True)
print("Tabla categories creada")
```

```python
schema_products = [
    bigquery.SchemaField("product_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("category_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("price", "FLOAT64", mode="REQUIRED"),
    bigquery.SchemaField("cost", "FLOAT64", mode="REQUIRED"),
    bigquery.SchemaField("stock", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("is_active", "BOOL", mode="REQUIRED"),
]
table = bigquery.Table(f"{dataset_id}.products", schema=schema_products)
client.create_table(table, exists_ok=True)
print("Tabla products creada")
```

```python
schema_orders = [
    bigquery.SchemaField("order_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("customer_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("status", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("shipping_country", "STRING", mode="NULLABLE"),
    bigquery.SchemaField("shipping_city", "STRING", mode="NULLABLE"),
    bigquery.SchemaField("order_date", "DATE", mode="REQUIRED"),
    bigquery.SchemaField("shipped_date", "DATE", mode="NULLABLE"),
    bigquery.SchemaField("delivered_date", "DATE", mode="NULLABLE"),
]
table = bigquery.Table(f"{dataset_id}.orders", schema=schema_orders)
client.create_table(table, exists_ok=True)
print("Tabla orders creada")
```

```python
schema_order_items = [
    bigquery.SchemaField("order_item_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("order_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("product_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("quantity", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("unit_price", "FLOAT64", mode="REQUIRED"),
    bigquery.SchemaField("discount", "FLOAT64", mode="NULLABLE"),
]
table = bigquery.Table(f"{dataset_id}.order_items", schema=schema_order_items)
client.create_table(table, exists_ok=True)
print("Tabla order_items creada")
```

```python
schema_payments = [
    bigquery.SchemaField("payment_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("order_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("payment_method", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("status", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("amount", "FLOAT64", mode="REQUIRED"),
    bigquery.SchemaField("payment_date", "DATE", mode="REQUIRED"),
]
table = bigquery.Table(f"{dataset_id}.payments", schema=schema_payments)
client.create_table(table, exists_ok=True)
print("Tabla payments creada")
```

```python
schema_reviews = [
    bigquery.SchemaField("review_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("order_item_id", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("rating", "INT64", mode="REQUIRED"),
    bigquery.SchemaField("comment", "STRING", mode="NULLABLE"),
    bigquery.SchemaField("review_date", "DATE", mode="REQUIRED"),
]
table = bigquery.Table(f"{dataset_id}.reviews", schema=schema_reviews)
client.create_table(table, exists_ok=True)
print("Tabla reviews creada")
```

```python
tables = client.list_tables(dataset_id)
print("Tablas en el dataset:")
for t in tables:
    print(" -", t.table_id)
```
